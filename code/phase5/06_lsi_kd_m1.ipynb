# Phase 5 (LSI) - Knowledge distillation: Model 1 (GQA) + LoRA student
The student is the frozen Phase-1 Model 1 with hand-written LoRA and the same BiGRU + attention head as the other runs. It is trained on
1. labelled documents (hard labels, same loss as the other runs: BCE with sqrt pos_weight, rare-statute oversampling), and
2. unlabelled test documents with the **soft labels of the best finished runs** (the teachers), averaged.

The teachers are the top runs by out-of-fold macro-F1 on the validation fold (picked automatically from the attached run outputs). The student never sees the validation fold, so its validation score is an honest comparison with the teachers. It also writes `oof_probs`, `test_probs` and a submission in the same format, so it can join the ensemble.

**Attach (Kaggle "Add Input"):**
1. The LSI data-prep output (contains `tokenized_lsi.pkl`, `lsi_utils.py`)
2. Your Phase-1 Model 1 folder (`legal-gqa-final`) and `legal_gqa_model.py`
3. The outputs of the finished runs that act as teachers (folders `lsi_m1_lora`, `lsi_m1_lorma`, `lsi_m4_lora`, `lsi_m4_lorma`; each needs `oof_probs_fold0.npy`, `oof_labels_fold0.npy`, `oof_doc_ids_fold0.json`, `test_probs_fold0.npy`)

**Output:** `/kaggle/working/lsi_kd_m1/` (checkpoints, OOF probabilities, test probabilities, submission)

## 3. Find the inputs

```python
def find_one_file(search_roots, file_name):
    for root in search_roots:
        for dir_path, dir_names, file_names in os.walk(root):
            if file_name in file_names:
                return os.path.join(dir_path, file_name)
    return None


def find_model_dir(name_hint):
    # The Model 1 folder. A Model 4 folder (legal-gqamoe-final, model_type legal_gqa_mha_moe) also contains
    # "legal-gqa" in its name, so it is skipped explicitly.
    root = os.environ.get("LSI_INPUT_ROOT", "/kaggle/input")
    for dir_path, dir_names, file_names in os.walk(root):
        low = dir_path.lower()
        if "config.json" not in file_names or name_hint not in low or "checkpoint" in low or "moe" in low:
            continue
        try:
            with open(os.path.join(dir_path, "config.json")) as fh:
                model_type = str(json.load(fh).get("model_type"))
        except Exception:
            model_type = ""
        if "moe" in model_type:
            continue
        return dir_path
    return None


def find_model_py(model_dir, file_name):
    # Prefer the copy that sits next to the model folder (same notebook output), then search everywhere.
    folder = os.path.dirname(os.path.abspath(model_dir))
    for _ in range(3):
        candidate = os.path.join(folder, file_name)
        if os.path.exists(candidate):
            return candidate
        folder = os.path.dirname(folder)
    return find_one_file([os.environ.get("LSI_INPUT_ROOT", "/kaggle/input"), "/kaggle/working"], file_name)


if DATA_DIR is None:
    pkl_path = find_one_file(["/kaggle/input", "/kaggle/working"], "tokenized_lsi.pkl")
    if pkl_path is None:
        raise FileNotFoundError("tokenized_lsi.pkl not found. Attach the LSI data-prep output.")
    DATA_DIR = os.path.dirname(pkl_path)

if MODEL_DIR is None:
    MODEL_DIR = find_model_dir(MODEL_NAME_HINT)
    if MODEL_DIR is None:
        raise FileNotFoundError("No Phase-1 Model 1 folder with config.json found. Attach legal-gqa-final.")

if MODEL_PY is None:
    MODEL_PY = find_model_py(MODEL_DIR, MODEL_PY_NAME)
    if MODEL_PY is None:
        raise FileNotFoundError("legal_gqa_model.py not found. Attach it (see the header).")

print("data dir:", DATA_DIR)
print("model dir:", MODEL_DIR)
print("model file:", MODEL_PY)
```

## 4. Load the data bundle from the data-prep notebook

```python
import pickle

with open(os.path.join(DATA_DIR, "tokenized_lsi.pkl"), "rb") as f:
    DATA = pickle.load(f)

if os.path.exists(os.path.join(DATA_DIR, "lsi_utils.py")):
    shutil.copy(os.path.join(DATA_DIR, "lsi_utils.py"), "lsi_utils.py")
sys.path.insert(0, os.getcwd())
import lsi_utils

SEP_ID = int(DATA["sep_id"])
PAD_ID = int(DATA["pad_id"])
STATUTES = [int(s) for s in DATA["statute_ids"]]
N_STATUTES = len(STATUTES)
print("SEP:", SEP_ID, "| PAD:", PAD_ID, "| statutes:", N_STATUTES)

# Pool = train + dev (test has no labels)
pool_ids, pool_labels, pool_fold, pool_doc_ids = [], [], [], []
for split in ["train", "dev"]:
    block = DATA["splits"][split]
    for i in range(len(block["doc_ids"])):
        pool_ids.append(block["ids"][i])
        pool_labels.append(block["labels"][i])
        pool_fold.append(int(block["fold"][i]))
        pool_doc_ids.append(str(block["doc_ids"][i]))

test_block = DATA["splits"]["test"]
test_ids = list(test_block["ids"])
test_doc_ids = [str(x) for x in test_block["doc_ids"]]

Y_POOL = lsi_utils.multi_hot(pool_labels, STATUTES)
N_POOL = len(pool_ids)

if SMOKE_TEST:
    keep_pool = list(range(min(SMOKE_DOCS, N_POOL)))
    pool_ids = [pool_ids[i] for i in keep_pool]
    pool_labels = [pool_labels[i] for i in keep_pool]
    pool_fold = [pool_fold[i] for i in keep_pool]
    pool_doc_ids = [pool_doc_ids[i] for i in keep_pool]
    Y_POOL = Y_POOL[keep_pool]
    N_POOL = len(pool_ids)
    test_ids = test_ids[:SMOKE_DOCS]
    test_doc_ids = test_doc_ids[:SMOKE_DOCS]

# Document sampling weights for rare-statute oversampling.
# A statute's rarity = sqrt(most common count / its count). A document gets the largest rarity of its statutes.
POOL_COUNTS = Y_POOL.sum(axis=0).astype(np.float64)
RARITY = np.sqrt(POOL_COUNTS.max() / np.maximum(POOL_COUNTS, 1.0))
DOC_WEIGHT = np.ones(N_POOL, dtype=np.float64)
for i in range(N_POOL):
    pos_idx = np.where(Y_POOL[i] == 1)[0]
    if len(pos_idx) > 0:
        DOC_WEIGHT[i] = float(RARITY[pos_idx].max())
DOC_WEIGHT = np.minimum(DOC_WEIGHT, RARE_OVERSAMPLE_CAP)
print("doc weight (min / mean / max):", round(float(DOC_WEIGHT.min()), 2),
      round(float(DOC_WEIGHT.mean()), 2), round(float(DOC_WEIGHT.max()), 2))

print("pool docs:", N_POOL, "| test docs:", len(test_ids))
print("fold counts:", {f: pool_fold.count(f) for f in sorted(set(pool_fold))})
print("label cardinality per pool doc (mean):", round(float(Y_POOL.sum(axis=1).mean()), 2))
```

## 4b. Teachers: pick the best finished runs and build the soft labels

```python
KD_FOLD = RUN_FOLDS[0]
if len(RUN_FOLDS) != 1:
    raise RuntimeError("The distillation run uses one fold. Set RUN_FOLDS to a single fold, for example [0].")
assert CKPT_EVERY % GRAD_ACC == 0, "CKPT_EVERY must be a multiple of GRAD_ACC"

INPUT_ROOT = os.environ.get("LSI_INPUT_ROOT", "/kaggle/input")
NEEDED_FILES = ["oof_probs_fold%d.npy" % KD_FOLD, "oof_labels_fold%d.npy" % KD_FOLD,
                "oof_doc_ids_fold%d.json" % KD_FOLD, "test_probs_fold%d.npy" % KD_FOLD]


def find_teacher_dir(name):
    found = []
    for dir_path, dir_names, file_names in os.walk(INPUT_ROOT):
        if os.path.basename(dir_path) != name:
            continue
        has_all = True
        for file_name in NEEDED_FILES:
            if file_name not in file_names:
                has_all = False
        if has_all:
            found.append(dir_path)
    if len(found) == 0:
        return None
    return sorted(found)[-1]


# The validation documents of the student's fold, in pool order.
val_rows_kd = []
val_ids_kd = []
for i in range(N_POOL):
    if pool_fold[i] == KD_FOLD:
        val_rows_kd.append(i)
        val_ids_kd.append(pool_doc_ids[i])
Y_val_kd = Y_POOL[val_rows_kd]

TEACHERS = []
for name in TEACHER_NAMES:
    folder = find_teacher_dir(name)
    if folder is None:
        print(name.ljust(14), ": not found in the inputs, skipped")
        continue
    with open(os.path.join(folder, NEEDED_FILES[2])) as fh:
        t_ids = [str(x) for x in json.load(fh)]
    t_probs = np.load(os.path.join(folder, NEEDED_FILES[0]))
    t_labels = np.load(os.path.join(folder, NEEDED_FILES[1]))
    t_test = np.load(os.path.join(folder, NEEDED_FILES[3]))

    row_of = {}
    for r in range(len(t_ids)):
        row_of[t_ids[r]] = r
    order = []
    for doc_id in val_ids_kd:
        if doc_id not in row_of:
            raise RuntimeError(name + ": validation document " + doc_id + " is missing from its OOF file. Same data bundle and folds needed.")
        order.append(row_of[doc_id])
    probs_aligned = t_probs[order]
    labels_aligned = t_labels[order]
    if not np.array_equal(labels_aligned.astype(np.int64), np.asarray(Y_val_kd).astype(np.int64)):
        raise RuntimeError(name + ": its OOF labels differ from this notebook's labels for the same documents.")
    if t_test.shape[0] != len(test_ids) or t_test.shape[1] != N_STATUTES:
        raise RuntimeError(name + ": test probabilities have shape " + str(t_test.shape) + ", expected " + str((len(test_ids), N_STATUTES)))

    best_t, best_macro = lsi_utils.best_global_threshold(probs_aligned, Y_val_kd)
    TEACHERS.append({"name": name, "dir": folder, "oof_macro_f1": float(best_macro), "threshold": float(best_t),
                     "oof": probs_aligned.astype(np.float64), "test": t_test.astype(np.float64)})
    print(name.ljust(14), "| OOF macro-F1 (best global threshold):", round(best_macro, 4), "|", folder)

if len(TEACHERS) == 0:
    raise RuntimeError("No teacher run found. Attach at least one finished run output (see the header).")


def teacher_score(t):
    return t["oof_macro_f1"]


TEACHERS = sorted(TEACHERS, key=teacher_score, reverse=True)
CHOSEN = TEACHERS[:N_TEACHERS]
print("teachers chosen:", [t["name"] for t in CHOSEN])

ens_oof = np.zeros_like(CHOSEN[0]["oof"])
TEACHER_TEST = np.zeros_like(CHOSEN[0]["test"])
for t in CHOSEN:
    ens_oof += t["oof"]
    TEACHER_TEST += t["test"]
ens_oof = ens_oof / len(CHOSEN)
TEACHER_TEST = TEACHER_TEST / len(CHOSEN)
TEACHER_ENS_T, TEACHER_ENS_MACRO = lsi_utils.best_global_threshold(ens_oof, Y_val_kd)
print("teacher ensemble OOF macro-F1 (the bar for the student):", round(TEACHER_ENS_MACRO, 4))

# Soft targets for the unlabelled documents, with the temperature applied to the teachers' logits.
clipped = np.clip(TEACHER_TEST, 1e-4, 1 - 1e-4)
teacher_logit = np.log(clipped / (1 - clipped))
TEACHER_SOFT = 1.0 / (1.0 + np.exp(-teacher_logit / KD_TEMP))
TEACHER_SOFT = TEACHER_SOFT.astype(np.float32)
print("soft labels:", TEACHER_SOFT.shape, "| mean predicted statutes per test doc at 0.5:",
      round(float((TEACHER_TEST >= 0.5).sum(axis=1).mean()), 2))

teacher_info = {"chosen": [t["name"] for t in CHOSEN],
                "all": [{"name": t["name"], "oof_macro_f1": t["oof_macro_f1"], "dir": t["dir"]} for t in TEACHERS],
                "teacher_ensemble_oof_macro_f1": float(TEACHER_ENS_MACRO), "kd_temp": KD_TEMP}
with open(os.path.join(OUT_DIR, "teacher_info.json"), "w") as f:
    json.dump(teacher_info, f, indent=1)
```

## 5. Load the frozen Phase-1 Model 1 backbone

```python
import importlib.util

spec = importlib.util.spec_from_file_location("legal_gqa_model", MODEL_PY)
legal_mod = importlib.util.module_from_spec(spec)
spec.loader.exec_module(legal_mod)

LM = legal_mod.LegalGQAForCausalLM.from_pretrained(MODEL_DIR)
LM = LM.to(DEVICE)


def find_backbone(lm):
    # The transformer without the LM head (the LM head would compute logits over 50k tokens for nothing).
    for name in ["model", "transformer", "backbone"]:
        if hasattr(lm, name):
            return getattr(lm, name)
    return lm


BACKBONE = find_backbone(LM)


def find_input_embedding(lm):
    # The largest nn.Embedding is the token embedding table.
    best_name = None
    best_module = None
    for name, module in lm.named_modules():
        if isinstance(module, nn.Embedding):
            if best_module is None or module.num_embeddings > best_module.num_embeddings:
                best_name = name
                best_module = module
    return best_name, best_module


def set_submodule(root, dotted_name, new_module):
    parts = dotted_name.split(".")
    parent = root
    for part in parts[:-1]:
        parent = getattr(parent, part)
    setattr(parent, parts[-1], new_module)


EMB_NAME, old_emb = find_input_embedding(BACKBONE)
print("token embedding:", EMB_NAME, "| rows:", old_emb.num_embeddings)

# PAD id is one past the tokenizer vocabulary, so the embedding table needs one extra row.
if PAD_ID >= old_emb.num_embeddings:
    new_emb = nn.Embedding(PAD_ID + 1, old_emb.embedding_dim).to(old_emb.weight.device, old_emb.weight.dtype)
    with torch.no_grad():
        new_emb.weight[: old_emb.num_embeddings] = old_emb.weight
        new_emb.weight[PAD_ID:] = 0.0      # the PAD row is always masked, so its value does not matter
    new_emb.weight.requires_grad = False
    set_submodule(BACKBONE, EMB_NAME, new_emb)
    print("embedding table grown from", old_emb.num_embeddings, "to", PAD_ID + 1, "rows (PAD row added)")
else:
    print("PAD id already inside the embedding table:", PAD_ID)

for p in BACKBONE.parameters():
    p.requires_grad = False
BACKBONE.eval()

n_frozen = sum(p.numel() for p in BACKBONE.parameters())
print("backbone:", type(BACKBONE).__name__, "| frozen params:", f"{n_frozen:,}")

# The transformer blocks: the ModuleList named blocks or layers.
BLOCK_LIST_NAME = None
BLOCK_LIST = None
for name, module in BACKBONE.named_modules():
    if isinstance(module, nn.ModuleList) and name.split(".")[-1] in ["blocks", "layers"]:
        BLOCK_LIST_NAME = name
        BLOCK_LIST = module
        break
if BLOCK_LIST is None:
    raise RuntimeError("No ModuleList named blocks or layers found. Check the Model 1 module names.")
print("transformer blocks:", BLOCK_LIST_NAME, "| count:", len(BLOCK_LIST))

# LoRA targets: linear layers inside the blocks only. Not the embedding, not lm_head.
LORA_TARGETS = []
for name, module in BACKBONE.named_modules():
    if isinstance(module, nn.Linear) and name.startswith(BLOCK_LIST_NAME + "."):
        LORA_TARGETS.append(name)
print("LoRA target linear layers:", len(LORA_TARGETS))
print("first few:", LORA_TARGETS[:6])

# Hidden states. The model's own forward runs, but a hook on the last block stops it there,
# so the 50k-way vocabulary projection is never computed. This is valid because the model is
# causal: right padding cannot change the outputs of real tokens.
class StopAfterBlock(Exception):
    pass


CAPTURED = {}


def last_block_hook(module, inputs, output):
    if isinstance(output, tuple):
        CAPTURED["h"] = output[0]
    else:
        CAPTURED["h"] = output
    raise StopAfterBlock()


LAST_BLOCK_HANDLE = BLOCK_LIST[-1].register_forward_hook(last_block_hook)

import inspect
LM_FORWARD_ARGS = list(inspect.signature(LM.forward).parameters.keys())
print("LM forward arguments:", LM_FORWARD_ARGS)
FORWARD_TAKES_MASK = "attention_mask" in LM_FORWARD_ARGS


def backbone_hidden(input_ids, attention_mask):
    # Returns the last block output [B, T, H].
    CAPTURED.clear()
    try:
        if FORWARD_TAKES_MASK:
            LM(input_ids=input_ids, attention_mask=attention_mask)
        else:
            LM(input_ids=input_ids)
    except StopAfterBlock:
        pass
    if "h" not in CAPTURED:
        raise RuntimeError("Hook did not capture the last block output.")
    return CAPTURED.pop("h")


HIDDEN = None
with torch.no_grad():
    probe_ids = torch.tensor([[1, 2, 3, SEP_ID]], device=DEVICE)
    probe_mask = torch.ones_like(probe_ids)
    probe_h = backbone_hidden(probe_ids, probe_mask)
    HIDDEN = int(probe_h.shape[-1])
print("backbone hidden size:", HIDDEN)
```

## 6. Hand-written LoRA layer
LoRA: `y = W0 x + (alpha / r) * B A x`. `W0` is frozen, `B` starts at zero, so training starts exactly at the Phase-1 weights.

```python
class LoRALinear(nn.Module):
    def __init__(self, base, r, alpha, dropout):
        super().__init__()
        self.base = base
        for p in self.base.parameters():
            p.requires_grad = False
        d_out, d_in = base.weight.shape
        self.r = r
        self.scale = alpha / r
        self.lora_A = nn.Parameter(torch.empty(r, d_in, device=base.weight.device, dtype=torch.float32))
        self.lora_B = nn.Parameter(torch.zeros(d_out, r, device=base.weight.device, dtype=torch.float32))
        self.drop = nn.Dropout(dropout)
        self.reset_parameters()

    def reset_parameters(self):
        nn.init.kaiming_uniform_(self.lora_A, a=math.sqrt(5))
        with torch.no_grad():
            self.lora_B.zero_()

    def forward(self, x):
        y = self.base(x)
        delta = self.drop(x.float()) @ self.lora_A.t() @ self.lora_B.t()
        return y + (delta * self.scale).to(y.dtype)


def set_submodule(root, dotted_name, new_module):
    parts = dotted_name.split(".")
    parent = root
    for part in parts[:-1]:
        parent = getattr(parent, part)
    setattr(parent, parts[-1], new_module)


def inject_lora(backbone):
    lora_modules = []
    for name in LORA_TARGETS:
        target = backbone.get_submodule(name)
        wrapped = LoRALinear(target, LORA_R, LORA_ALPHA, LORA_DROPOUT)
        set_submodule(backbone, name, wrapped)
        lora_modules.append(wrapped)
    return lora_modules


# Identity check: LoRA at init must not change the Phase-1 hidden states.
check_ids = torch.tensor([[5, 9, 12, 40, 77, SEP_ID]], device=DEVICE)
check_mask = torch.ones_like(check_ids)
with torch.no_grad():
    before = backbone_hidden(check_ids, check_mask).float().clone()

LORA_MODULES = inject_lora(BACKBONE)
BACKBONE.eval()
for m in LORA_MODULES:
    m.train()   # keep LoRA dropout on during training only

with torch.no_grad():
    after = backbone_hidden(check_ids, check_mask).float()
identity_diff = float((before - after).abs().max())
print("LoRA modules:", len(LORA_MODULES))
print("identity check (max abs diff at init, should be 0):", identity_diff)
assert identity_diff < 1e-4, "LoRA changed the Phase-1 output at init"

LORA_PARAMS = []
for m in LORA_MODULES:
    LORA_PARAMS.append(m.lora_A)
    LORA_PARAMS.append(m.lora_B)
n_lora = sum(p.numel() for p in LORA_PARAMS)
print("trainable LoRA params:", f"{n_lora:,}")


def reset_lora():
    for m in LORA_MODULES:
        m.reset_parameters()
```

## 7. Chunk-level encoder and document head

```python
class DocHead(nn.Module):
    # Chunk vectors -> BiGRU -> attention pooling over chunks -> one logit per statute.
    def __init__(self, in_dim, hidden, n_out, dropout=0.1):
        super().__init__()
        self.gru = nn.GRU(in_dim, hidden, batch_first=True, bidirectional=True)
        self.att = nn.Linear(2 * hidden, 1)
        self.drop = nn.Dropout(dropout)
        self.out = nn.Linear(2 * hidden, n_out)

    def forward(self, chunk_vecs, chunk_mask):
        h, _ = self.gru(chunk_vecs)
        scores = self.att(h).squeeze(-1)
        scores = scores.masked_fill(chunk_mask == 0, -1e4)
        weights = torch.softmax(scores, dim=1)
        doc = (weights.unsqueeze(-1) * h).sum(dim=1)
        return self.out(self.drop(doc))


HEAD = DocHead(HIDDEN + 1, HEAD_HIDDEN, N_STATUTES).to(DEVICE)
print("head params:", f"{sum(p.numel() for p in HEAD.parameters()):,}")


def pooled_backbone(ids_t, mask_t):
    # Backbone forward, then mean pooling over real (non-pad) tokens. Returns [N, HIDDEN].
    with torch.autocast(device_type=DEVICE.type, dtype=AMP_DTYPE, enabled=USE_AMP):
        hidden = backbone_hidden(ids_t, mask_t)
    hidden = hidden.float()
    m = mask_t.unsqueeze(-1).float()
    return (hidden * m).sum(dim=1) / m.sum(dim=1).clamp(min=1.0)


def encode_chunks(chunk_list, rel_list):
    # chunk_list: list of id lists (each ends with SEP). Returns [N, HIDDEN + 1] (mean pool + position).
    ids, mask, lengths = lsi_utils.pad_right(chunk_list, PAD_ID)
    ids_t = torch.tensor(ids, device=DEVICE)
    mask_t = torch.tensor(mask, device=DEVICE)
    if torch.is_grad_enabled():
        # Gradient checkpointing: this micro-batch's activations are recomputed during backward
        # instead of being kept in memory. Needed on a 16 GB T4 with 32 chunks per document.
        mean_vec = torch_checkpoint(pooled_backbone, ids_t, mask_t, use_reentrant=False)
    else:
        mean_vec = pooled_backbone(ids_t, mask_t)
    rel = torch.tensor(rel_list, device=DEVICE, dtype=torch.float32).unsqueeze(-1)
    return torch.cat([mean_vec, rel], dim=-1)


def chunks_from_starts(ids, starts):
    # Build chunk id lists from start offsets. Each chunk ends with SEP.
    ids = np.asarray(ids)
    n = len(ids)
    body_len = CHUNK_LEN - 1
    chunks = []
    rel = []
    for s in starts:
        piece = ids[s:s + body_len].tolist()
        piece.append(int(SEP_ID))
        chunks.append(piece)
        rel.append(s / max(1, n))
    return chunks, rel


def build_doc_features(ids_list, max_chunks):
    # ids_list: list of documents (each a numpy array of token ids).
    # Returns padded chunk features [B, C, HIDDEN+1] and chunk mask [B, C].
    flat_chunks = []
    flat_rel = []
    counts = []
    for ids in ids_list:
        # training and inference use the same chunking (same cap, same pick)
        chunks, rel = lsi_utils.make_chunks(ids, SEP_ID, CHUNK_LEN, CHUNK_OVERLAP, max_chunks, CHUNK_PICK)
        counts.append(len(chunks))
        flat_chunks.extend(chunks)
        flat_rel.extend(rel)

    pieces = []
    for start in range(0, len(flat_chunks), MICRO_CHUNKS):
        pieces.append(encode_chunks(flat_chunks[start:start + MICRO_CHUNKS], flat_rel[start:start + MICRO_CHUNKS]))
    flat_vec = torch.cat(pieces, dim=0)

    c_max = max(counts)
    batch = len(ids_list)
    padded = torch.zeros(batch, c_max, HIDDEN + 1, device=DEVICE, dtype=flat_vec.dtype)
    cmask = torch.zeros(batch, c_max, device=DEVICE)
    pos = 0
    for i in range(batch):
        c = counts[i]
        padded[i, :c] = flat_vec[pos:pos + c]
        cmask[i, :c] = 1.0
        pos += c
    return padded, cmask
```

## 8. Loss, metrics, and checkpoints

```python
def make_pos_weight():
    raw = DATA["pos_weight"]
    values = []
    for s in STATUTES:
        w = float(raw.get(s, raw.get(str(s), 1.0)))
        if POS_WEIGHT_MODE == "sqrt":
            w = math.sqrt(w)
        values.append(w)
    return torch.tensor(values, dtype=torch.float32, device=DEVICE)


POS_WEIGHT = make_pos_weight()
LOSS_FN = nn.BCEWithLogitsLoss(pos_weight=POS_WEIGHT)
LOSS_FN_NONE = nn.BCEWithLogitsLoss(pos_weight=POS_WEIGHT, reduction="none")
print("pos_weight mode:", POS_WEIGHT_MODE, "| min:", round(float(POS_WEIGHT.min()), 2),
      "| max:", round(float(POS_WEIGHT.max()), 2))


def time_left_sec():
    return SESSION_CAP_SEC - (time.time() - START_TIME)


def ckpt_path(fold):
    return os.path.join(OUT_DIR, "ckpt_fold%d.pt" % fold)


def save_ckpt(fold, epoch_done, micro_done, optimizer, scheduler, scaler):
    # epoch_done = finished epochs; micro_done = micro-steps already done inside the next epoch (a multiple of GRAD_ACC).
    state = {
        "fold": fold,
        "epoch_done": epoch_done,
        "micro_done": micro_done,
        "lora": [(m.lora_A.detach().cpu(), m.lora_B.detach().cpu()) for m in LORA_MODULES],
        "head": HEAD.state_dict(),
        "optimizer": optimizer.state_dict(),
        "scheduler": scheduler.state_dict(),
        "scaler": scaler.state_dict() if scaler is not None else None,
    }
    torch.save(state, ckpt_path(fold))
    print("checkpoint saved: fold", fold, "| epochs done:", epoch_done, "| micro-steps into the next epoch:", micro_done)


def load_ckpt(fold, optimizer, scheduler, scaler):
    if not os.path.exists(ckpt_path(fold)):
        return 0, 0
    state = torch.load(ckpt_path(fold), map_location="cpu")
    for m, (a, b) in zip(LORA_MODULES, state["lora"]):
        m.lora_A.data.copy_(a.to(DEVICE))
        m.lora_B.data.copy_(b.to(DEVICE))
    HEAD.load_state_dict(state["head"])
    optimizer.load_state_dict(state["optimizer"])
    scheduler.load_state_dict(state["scheduler"])
    if scaler is not None and state["scaler"] is not None:
        scaler.load_state_dict(state["scaler"])
    print("resumed fold", fold, "from epoch", state["epoch_done"], "+", state.get("micro_done", 0), "micro-steps")
    return int(state["epoch_done"]), int(state.get("micro_done", 0))


def best_label_thresholds(P, Y, global_t, min_pos):
    # One threshold per statute: the one that maximises that statute's F1.
    # Statutes with fewer than min_pos positives keep the global threshold.
    grid = np.arange(0.05, 0.951, 0.01)
    thr = np.full(P.shape[1], float(global_t))
    for j in range(P.shape[1]):
        pos = int(Y[:, j].sum())
        if pos < min_pos:
            continue
        best_f = -1.0
        best_t = float(global_t)
        for t in grid:
            pred = P[:, j] >= t
            tp = int((pred & (Y[:, j] == 1)).sum())
            fp = int((pred & (Y[:, j] == 0)).sum())
            fn = pos - tp
            denom = 2 * tp + fp + fn
            f = (2.0 * tp / denom) if denom > 0 else 0.0
            if f > best_f:
                best_f = f
                best_t = float(t)
        thr[j] = best_t
    return thr


def macro_micro_subset(Y_true, Y_pred):
    return lsi_utils.macro_micro_f1(Y_true, Y_pred)


@torch.no_grad()
def predict_probs(ids_all, max_chunks):
    LORA_MODULES_EVAL = LORA_MODULES
    for m in LORA_MODULES_EVAL:
        m.eval()
    HEAD.eval()
    out = []
    for start in range(0, len(ids_all), EVAL_BATCH):
        docs = ids_all[start:start + EVAL_BATCH]
        feats, cmask = build_doc_features(docs, max_chunks)
        logits = HEAD(feats.float(), cmask)
        out.append(torch.sigmoid(logits).cpu().numpy())
    for m in LORA_MODULES_EVAL:
        m.train()
    return np.concatenate(out, axis=0)
```

## 9. Train one fold (resumable, with a session time cap)

```python
def make_epoch_order(fold, epoch, train_rows):
    # The same (fold, epoch) always gives the same document order, so a run can resume in the middle of an epoch.
    # Items are (kind, index): kind 0 = labelled pool document, kind 1 = unlabelled test document.
    rng = np.random.RandomState(SEED + 1000 * fold + epoch)
    n_lab = SMOKE_LAB if SMOKE_TEST else N_LAB_PER_EPOCH
    n_unl = SMOKE_UNL if SMOKE_TEST else N_UNL_PER_EPOCH
    if RARE_OVERSAMPLE:
        w = DOC_WEIGHT[train_rows]
        w = w / w.sum()
        lab_draw = rng.choice(train_rows, size=n_lab, replace=True, p=w)
    else:
        lab_draw = rng.choice(train_rows, size=n_lab, replace=True)
    items = []
    for row in lab_draw:
        items.append((0, int(row)))
    if USE_TEST_SOFT:
        n_unl = min(n_unl, len(test_ids))
        unl_draw = rng.choice(len(test_ids), size=n_unl, replace=False)
        for idx in unl_draw:
            items.append((1, int(idx)))
    perm = rng.permutation(len(items))
    shuffled = []
    for k in perm:
        shuffled.append(items[k])
    return shuffled


def train_one_fold(fold):
    train_rows = [i for i in range(N_POOL) if pool_fold[i] != fold]
    val_rows = [i for i in range(N_POOL) if pool_fold[i] == fold]
    if SMOKE_TEST:
        train_rows = train_rows[:SMOKE_DOCS]
        val_rows = val_rows[:SMOKE_DOCS]
    epochs = SMOKE_EPOCHS if SMOKE_TEST else EPOCHS

    reset_lora()
    HEAD.apply(lambda m: m.reset_parameters() if hasattr(m, "reset_parameters") else None)

    params = LORA_PARAMS + list(HEAD.parameters())
    optimizer = torch.optim.AdamW(params, lr=LR, weight_decay=WEIGHT_DECAY)

    n_items = len(make_epoch_order(fold, 0, train_rows))
    steps_per_epoch = math.ceil(n_items / (MICRO_BATCH * GRAD_ACC))
    total_steps = max(1, steps_per_epoch * epochs)
    warmup = max(1, int(WARMUP_FRAC * total_steps))

    def lr_lambda(step):
        if step < warmup:
            return float(step + 1) / warmup
        progress = (step - warmup) / max(1, total_steps - warmup)
        return 0.5 * (1.0 + math.cos(math.pi * min(1.0, progress)))

    scheduler = torch.optim.lr_scheduler.LambdaLR(optimizer, lr_lambda)
    scaler = torch.amp.GradScaler("cuda") if USE_SCALER else None

    epoch_start, micro_start = load_ckpt(fold, optimizer, scheduler, scaler)
    if EVAL_ONLY:
        if not os.path.exists(ckpt_path(fold)):
            raise RuntimeError("EVAL_ONLY is on, but no checkpoint for fold %d was found. Attach the earlier run's output "
                               "(a folder named like OUT_DIR with ckpt_fold%d.pt) or set RESUME_DIR." % (fold, fold))
        epochs = epoch_start          # no more training: predict with what was trained so far
        steps_done = int(scheduler.last_epoch)
        print("EVAL_ONLY: no training. Finished epochs in the checkpoint:", epoch_start,
              "| optimizer steps done:", steps_done, "of", steps_per_epoch, "per epoch")

    for epoch in range(epoch_start, epochs):
        order = make_epoch_order(fold, epoch, train_rows)
        micro_steps = math.ceil(len(order) / MICRO_BATCH)
        first_micro = micro_start if epoch == epoch_start else 0
        HEAD.train()
        for m in LORA_MODULES:
            m.train()
        running = 0.0
        n_seen = 0
        t0 = time.time()
        optimizer.zero_grad()
        for micro in range(first_micro, micro_steps):
            if time_left_sec() < 300:
                safe_micro = (micro // GRAD_ACC) * GRAD_ACC     # back to the last finished optimizer step
                save_ckpt(fold, epoch, safe_micro, optimizer, scheduler, scaler)
                print("session time cap reached. Run this notebook again (with this run's output attached) to resume.")
                return None
            items = order[micro * MICRO_BATCH:(micro + 1) * MICRO_BATCH]
            ids_list = []
            y_hard = np.zeros((len(items), N_STATUTES), dtype=np.float32)
            y_soft = np.zeros((len(items), N_STATUTES), dtype=np.float32)
            is_lab = np.zeros(len(items), dtype=bool)
            for j in range(len(items)):
                kind, idx = items[j]
                if kind == 0:
                    ids_list.append(pool_ids[idx])
                    y_hard[j] = Y_POOL[idx]
                    is_lab[j] = True
                else:
                    ids_list.append(test_ids[idx])
                    y_soft[j] = TEACHER_SOFT[idx]
            y_hard_t = torch.tensor(y_hard, device=DEVICE)
            y_soft_t = torch.tensor(y_soft, device=DEVICE)
            is_lab_t = torch.tensor(is_lab, device=DEVICE)
            with torch.autocast(device_type=DEVICE.type, dtype=AMP_DTYPE, enabled=USE_AMP):
                feats, cmask = build_doc_features(ids_list, MAX_CHUNKS_TRAIN)
                logits = HEAD(feats.float(), cmask)
            logits = logits.float()
            hard_loss = LOSS_FN_NONE(logits, y_hard_t).mean(dim=1)
            soft_loss = F.binary_cross_entropy_with_logits(logits / KD_TEMP, y_soft_t, reduction="none").mean(dim=1)
            soft_loss = soft_loss * (KD_TEMP * KD_TEMP) * KD_WEIGHT
            per_doc = torch.where(is_lab_t, hard_loss, soft_loss)
            loss = per_doc.mean() / GRAD_ACC
            if scaler is not None:
                scaler.scale(loss).backward()
            else:
                loss.backward()
            running += float(loss.detach()) * GRAD_ACC * len(items)
            n_seen += len(items)

            last_micro = (micro == micro_steps - 1)
            if (micro + 1) % GRAD_ACC == 0 or last_micro:
                if scaler is not None:
                    scaler.unscale_(optimizer)
                torch.nn.utils.clip_grad_norm_(params, GRAD_CLIP)
                if scaler is not None:
                    scaler.step(optimizer)
                    scaler.update()
                else:
                    optimizer.step()
                scheduler.step()
                optimizer.zero_grad()
                if (micro + 1) % CKPT_EVERY == 0 and not last_micro:
                    save_ckpt(fold, epoch, micro + 1, optimizer, scheduler, scaler)

            if (micro - first_micro) % 500 == 0:
                print("epoch", epoch + 1, "| micro", micro, "/", micro_steps,
                      "| loss", round(running / max(1, n_seen), 5),
                      "| elapsed", round(time.time() - t0, 1), "s")

        print("epoch", epoch + 1, "done | train loss", round(running / max(1, n_seen), 5),
              "| seconds", round(time.time() - t0, 1))
        save_ckpt(fold, epoch + 1, 0, optimizer, scheduler, scaler)

    # validation on the held-out fold, all chunks
    val_ids = [pool_ids[i] for i in val_rows]
    val_probs = predict_probs(val_ids, MAX_CHUNKS_EVAL)
    val_true = Y_POOL[val_rows]
    macro05, micro05, subset05 = macro_micro_subset(val_true, (val_probs >= 0.5).astype(int))
    best_t, best_macro = lsi_utils.best_global_threshold(val_probs, val_true)
    print("fold", fold, "| macro-F1 @0.5:", round(macro05, 4), "| micro-F1 @0.5:", round(micro05, 4),
          "| subset acc @0.5:", round(subset05, 4))
    print("fold", fold, "| student: best global threshold:", round(best_t, 2), "| macro-F1 at that threshold:", round(best_macro, 4))
    print("fold", fold, "| teacher ensemble on the same documents:", round(TEACHER_ENS_MACRO, 4),
          "| best single teacher:", round(CHOSEN[0]["oof_macro_f1"], 4))

    np.save(os.path.join(OUT_DIR, "oof_probs_fold%d.npy" % fold), val_probs.astype(np.float32))
    with open(os.path.join(OUT_DIR, "oof_doc_ids_fold%d.json" % fold), "w") as f:
        json.dump([pool_doc_ids[i] for i in val_rows], f)
    np.save(os.path.join(OUT_DIR, "oof_labels_fold%d.npy" % fold), val_true.astype(np.uint8))

    # test predictions for this fold's model
    test_probs = predict_probs(test_ids, MAX_CHUNKS_EVAL)
    np.save(os.path.join(OUT_DIR, "test_probs_fold%d.npy" % fold), test_probs.astype(np.float32))
    print("saved OOF and test probabilities for fold", fold)

    return {"fold": fold, "macro_f1_0.5": macro05, "micro_f1_0.5": micro05,
            "subset_acc_0.5": subset05, "best_threshold": best_t, "best_macro_f1": best_macro,
            "epochs_trained": epochs}


RESULTS = []
for fold in RUN_FOLDS:
    print("\n=== fold", fold, "===")
    result = train_one_fold(fold)
    if result is None:
        break
    RESULTS.append(result)

print("\nfolds finished:", [r["fold"] for r in RESULTS])
```

## 10. Combine folds, pick a threshold, write the submission

```python
done_folds = [r["fold"] for r in RESULTS]
if len(done_folds) > 0:
    # threshold chosen on the out-of-fold probabilities of the folds that finished
    oof_list, label_list = [], []
    for f in done_folds:
        oof_list.append(np.load(os.path.join(OUT_DIR, "oof_probs_fold%d.npy" % f)))
        label_list.append(np.load(os.path.join(OUT_DIR, "oof_labels_fold%d.npy" % f)))
    oof_probs = np.concatenate(oof_list, axis=0)
    oof_true = np.concatenate(label_list, axis=0).astype(np.int64)
    global_t, global_macro = lsi_utils.best_global_threshold(oof_probs, oof_true)
    print("OOF rows:", oof_probs.shape[0], "| global threshold:", round(global_t, 2),
          "| OOF macro-F1:", round(global_macro, 4))

    # Honest check: tune per-label thresholds on half the OOF rows, score on the other half.
    half = (np.arange(oof_probs.shape[0]) % 2) == 0
    thr_a = best_label_thresholds(oof_probs[half], oof_true[half], global_t, THRESH_MIN_POS)
    thr_b = best_label_thresholds(oof_probs[~half], oof_true[~half], global_t, THRESH_MIN_POS)
    glob_on_b = lsi_utils.macro_micro_f1(oof_true[~half], (oof_probs[~half] >= global_t).astype(int))[0]
    per_on_b = lsi_utils.macro_micro_f1(oof_true[~half], (oof_probs[~half] >= thr_a).astype(int))[0]
    glob_on_a = lsi_utils.macro_micro_f1(oof_true[half], (oof_probs[half] >= global_t).astype(int))[0]
    per_on_a = lsi_utils.macro_micro_f1(oof_true[half], (oof_probs[half] >= thr_b).astype(int))[0]
    print("held-out check | global threshold macro-F1:", round((glob_on_a + glob_on_b) / 2, 4),
          "| per-label thresholds macro-F1:", round((per_on_a + per_on_b) / 2, 4))

    # final thresholds use all OOF rows
    thr_all = best_label_thresholds(oof_probs, oof_true, global_t, THRESH_MIN_POS)
    final_pred = (oof_probs >= thr_all).astype(int)
    final_macro, final_micro, final_subset = lsi_utils.macro_micro_f1(oof_true, final_pred)
    print("final per-label thresholds on all OOF | macro-F1:", round(final_macro, 4),
          "| micro-F1:", round(final_micro, 4), "| subset acc:", round(final_subset, 4))
    with open(os.path.join(OUT_DIR, "per_label_thresholds.json"), "w") as f:
        json.dump({str(STATUTES[j]): float(thr_all[j]) for j in range(N_STATUTES)}, f, indent=1)

    test_avg = np.mean([np.load(os.path.join(OUT_DIR, "test_probs_fold%d.npy" % f)) for f in done_folds], axis=0)

    pred_lists = []
    for i in range(test_avg.shape[0]):
        row = test_avg[i]
        chosen = [STATUTES[j] for j in range(N_STATUTES) if row[j] >= thr_all[j]]
        if len(chosen) == 0:
            # every training document has at least one statute, so always predict the top one
            chosen = [STATUTES[int(np.argmax(row))]]
        pred_lists.append(chosen)

    sub_path = os.path.join(OUT_DIR, "submission_lsi_kd_m1.json")
    lsi_utils.write_submission(test_doc_ids, pred_lists, STATUTES, sub_path)

    summary = {
        "model": "Phase-1 Model 1 (GQA) + LoRA, distilled from the best finished runs",
        "lora": {"r": LORA_R, "alpha": LORA_ALPHA, "dropout": LORA_DROPOUT, "targets": len(LORA_TARGETS)},
        "folds": RESULTS,
        "kd": {"teachers": teacher_info["chosen"], "teacher_ensemble_oof_macro_f1": teacher_info["teacher_ensemble_oof_macro_f1"],
               "use_test_soft": USE_TEST_SOFT, "kd_temp": KD_TEMP, "kd_weight": KD_WEIGHT,
               "n_lab_per_epoch": N_LAB_PER_EPOCH, "n_unl_per_epoch": N_UNL_PER_EPOCH},
        "oof_macro_f1_global_threshold": global_macro,
        "global_threshold": global_t,
        "oof_macro_f1_per_label_thresholds": final_macro,
        "held_out_check": {"global": (glob_on_a + glob_on_b) / 2, "per_label": (per_on_a + per_on_b) / 2},
        "config": {"chunk_len": CHUNK_LEN, "overlap": CHUNK_OVERLAP,
                   "max_chunks_train": MAX_CHUNKS_TRAIN, "max_chunks_eval": MAX_CHUNKS_EVAL,
                   "lr": LR, "epochs": EPOCHS, "pos_weight_mode": POS_WEIGHT_MODE,
                   "effective_batch": MICRO_BATCH * GRAD_ACC, "seed": SEED, "eval_only": EVAL_ONLY},
    }
    with open(os.path.join(OUT_DIR, "run_summary.json"), "w") as f:
        json.dump(summary, f, indent=1)
    print("run summary saved")
else:
    print("No fold finished yet. Run again to resume.")

print("elapsed (min):", round((time.time() - START_TIME) / 60, 1))
```

## Notes
- Student = Phase-1 Model 1 + LoRA. Teachers = the best finished runs by out-of-fold macro-F1 (picked automatically). Soft labels exist only for the test documents (the finished runs saved probabilities for the validation fold and the test set, not for the training documents), so the soft loss is used on unlabelled test documents and the hard loss on labelled documents.
- This uses the unlabelled test inputs (never their labels), as in the Phase 3 distillation. Set `USE_TEST_SOFT = False` to turn that off; the run then becomes plain training.
- An epoch is `N_LAB_PER_EPOCH + N_UNL_PER_EPOCH` documents (about 30k, a bit under one full pass of the other runs). Time one with `SMOKE_TEST = True` first, then adjust.
- A checkpoint is saved every `CKPT_EVERY` micro-steps, including inside an epoch. If the session cap stops the run, attach its output to a new run (the folder `lsi_kd_m1` is found automatically) and run again to continue. `EVAL_ONLY = True` skips training and predicts from the checkpoint.
- Set `MICRO_BATCH = 1` if the GPU runs out of memory.
